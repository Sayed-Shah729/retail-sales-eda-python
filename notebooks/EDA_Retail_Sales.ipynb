## Region With Highest and Lowest Sale

```python
regions_sales = data.groupby("Region")["Sales"].sum()
regions_sales
```

```python
print("Region with Highest Sales: ", regions_sales.idxmax())
```

```python
print("Region with lowest Sales: ", regions_sales.idxmin())
```

# Part C — Grouping and Simple Analysis

```python
print("Total_sales_and_Total_profit_by_category ")
data.groupby("Category")[["Sales", "Profit"]].sum()
```

```python
print("Average Profit and Discount By subcategory")
data.groupby("Sub-Category")[["Profit", "Discount"]].mean()
```

## Top 5 cities with highest sales


```python
top_cities = data.groupby("City")["Sales"].sum().sort_values(ascending=False).head(5)
print("Top 5 Cities with Highest Sales: ", top_cities)
```

```python
print("Profit by region and segmemt")
data.groupby(["Region", "Segment"])["Profit"].sum()
```

# Part D — Visualizations

```python
data.groupby("Category")["Sales"].sum().plot(kind="bar" , color ="skyblue")
plt.title("Sales By Category")
plt.ylabel("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Pie chart - sales by region
data.groupby("Region")["Sales"].sum().plot(kind="pie",  startangle=90)
plt.title("Sales Share by Region")
plt.ylabel("")
plt.show()
```

```python
import matplotlib.pyplot as plt
# Histogram - distribution of profit
plt.hist(data["Profit"], bins=20, color='orange', edgecolor='black')
plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.ylabel("Frequency")
plt.show()
```

```python
import seaborn as sns
```

```python
# Boxplot - Sales by Ship Mode
sns.boxplot(x="Ship Mode", y="Sales", data=data)
plt.title("Sales by Ship Mode")
plt.show()
```

```python
import matplotlib.pyplot as plt
# Scatter plot - Sales vs Profit
plt.scatter(data["Sales"], data["Profit"], color='green')
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
import matplotlib.pyplot as plt
# Convert Order Date to datetime and plot monthly sales
data["Order Date"] = pd.to_datetime(data["Order Date"])
data["Month"] = data["Order Date"].dt.to_period("M")
monthly_sales = data.groupby("Month")["Sales"].sum()
monthly_sales.plot(kind="line", marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.show()
```

# Part E — Deeper Analysis

```python
# Correlation heatmap
corr = data[["Sales", "Profit", "Discount", "Quantity"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()







```

```python
# Orders with negative profit
loss_orders = data[data["Profit"] < 0]
print("Number of loss orders:", len(loss_orders))
print("Percentage of loss orders:", (len(loss_orders) / len(data)) * 100, "%")
```

```python
# Top 5 customers by sales
top_customers = data.groupby("Customer Name")["Sales"].sum().sort_values(ascending=False).head(5)
print("\nTop 5 Customers:\n", top_customers)
```

```python
# Top 5 profitable products
top_products = data.groupby("Product Name")["Profit"].sum().sort_values(ascending=False).head(5)
print("\nTop 5 Profitable Products:\n", top_products)
```

```python
# Month with highest and lowest sales
best_month = monthly_sales.idxmax()
worst_month = monthly_sales.idxmin()
print("\nHighest Sales Month:", best_month)
print("Lowest Sales Month:", worst_month)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Trend line
sns.lineplot(x=monthly_sales.index.astype(str), y=monthly_sales.values, marker='o')
plt.xticks(rotation=45)
plt.title("Sales Trend Over Time")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.show()
```
